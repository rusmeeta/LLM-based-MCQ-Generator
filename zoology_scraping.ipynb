```python
import pymupdf

doc = pymupdf.open("Zoology.pdf")
print("Total pages:", len(doc))

text = ""

for pno in range(20, 57):
    page = doc[pno]

    page_h = page.rect.height

    for block in page.get_text("dict")["blocks"]:
        if block["type"] != 0:
            continue
        for line in block["lines"]:
            line_text = ""
            for span in line["spans"]:
                line_text += span["text"]
            line_text = " ".join(line_text.split())

            if not line_text:
                continue

            x0, y0, x1, y1 = line["bbox"]

            # Position: header band
            if y0 < 0.08 * page_h:
                continue
            # Position: footer band
            if y1 > 0.92 * page_h:
                continue

            # Figure caption
            if line_text.startswith("Figure"):
                continue


            text += line_text + "\n"

doc.close()

# Whitespace cleanup
lines = [" ".join(l.split()) for l in text.splitlines() if l.strip()]
cleaned = "\n".join(lines)
print(cleaned[:4000])
```

```python
cleaned = []
for line in lines:
    s = line.strip()

    # Drop bold-only running header
    if (s.startswith("**") and s.endswith("**")
        and s.count("**") == 2
        and not s.startswith("#")
        and not any(c.isdigit() for c in s)):
        continue

    if s.startswith("Figure ") and len(s) < 100:
        continue

    if line.strip() == "Human Anatomy and Physiology - I" :
        continue
         
    if line.strip() == "Zoology":
        continue

    cleaned.append(line)

lines = cleaned
print("After bold-header step:", len(lines), "lines")
print("\n".join(lines[:60]))
```

```python
cleaned = []
for line in lines:
    s = line

    # Remove heading markers (### at start)
    while s.startswith("#"):
        s = s[1:]
    s = s.lstrip()   # remove leading space

    # Remove bold markers (**...**)
    s = s.replace("**", "")

    # Remove italic markers (_..._)
    s = s.replace("_", "")

    # Remove stray asterisks (*)
    s = s.replace("*", "")

    cleaned.append(s)

lines = cleaned
print("After removing markers:", len(lines), "lines")
print("\n".join(lines[:40]))

with open("chapter1_clean.md", "w", encoding="utf-8") as f:
    f.write("\n".join(lines))

print("Saved chapter1_clean.md")
print("Lines:", len(lines))
print("Chars:", len("\n".join(lines)))
```

```python
import os
import json
from dotenv import load_dotenv
from google import genai
from typing import Literal
from pydantic import BaseModel, Field
from google.genai import types
import time

# 1. Load API key
load_dotenv(".env")
api_key = os.environ.get("GEMINI_API_KEY")

if not api_key:
    raise ValueError("GEMINI_API_KEY not found. Check your .env file.")
    
client = genai.Client(api_key=api_key)


SYSTEM_PROMPT = """You are an expert exam question writer for a Zoology textbook.

Your task is to create high-quality multiple-choice questions (MCQs)
based ONLY on the source text provided to you.

==================================================
1. SOURCE GROUNDING
==================================================

Every question, correct answer, and distractor must be supported by
information contained in the provided source text.

Do NOT introduce:
- General scientific knowledge that is not present in the source
- Information from outside the source
- Modern medical/scientific findings not stated in the source
- Historical context that is not stated in the source
- Comparisons that are not stated or clearly supported by the source
- Causes, effects, purposes, or mechanisms that the source does not establish

If a question cannot be answered using ONLY the provided source,
do not generate it.

Do not invent scientific facts simply to create questions or distractors.


==================================================
2. NUMBER OF QUESTIONS
==================================================

First identify the distinct, testable concepts present in the source.

- If the source is only a heading, title, label, or contains no
  testable information → return 0 questions.

- If the source contains 1-3 distinct testable concepts → generate
  approximately 1-3 questions.

- If the source contains several distinct testable concepts → generate
  as many high-quality questions as the source reasonably supports,
  usually up to 10 questions.

- Do NOT force a fixed number of questions.

- Do NOT generate a question merely to increase the number of MCQs.

- Do NOT combine unrelated facts just to create another question.

- Quality, source support, and conceptual coverage are more important
  than reaching a target number.

- Do not skip a strong testable concept merely because enough questions
  have already been generated.

- Avoid generating multiple questions that test exactly the same fact
  unless the source supports genuinely different aspects of that fact.

If the text has nothing worth testing, return:

{"mcqs": []}


==================================================
3. IDENTIFY TESTABLE CONCEPTS
==================================================

Before writing questions, identify the distinct concepts or facts that
can reasonably be tested.

Examples of testable concepts:

- Definitions
- Functions
- Characteristics
- Causes
- Effects
- Processes
- Structures
- Classifications
- Differences between related concepts
- Anatomical relationships
- Biological mechanisms
- Examples explicitly given in the source
- Sequences or stages
- Scientific properties

Do not treat every sentence as a separate fact.

Prefer meaningful concepts over superficial wording differences.


==================================================
4. QUESTION QUALITY
==================================================

Questions should test understanding of the source rather than simply
copying a sentence from it.

Questions may test:

- Direct recall
- Identification
- Distinction between related concepts
- Explanation
- Understanding of a process
- Application of concepts

Do NOT artificially make simple facts complicated.

A simple factual concept may have a simple question.

The difficulty should come from the concept being tested, not from
unnecessarily complicated vocabulary or sentence structure.


==================================================
5. NATURAL QUESTION WORDING
==================================================

Write questions in clear, natural language appropriate for a Zoology
student.

Prefer direct question stems such as:

- What is the function of...?
- Which feature characterizes...?
- Why does...?
- Which of the following...?
- How does...?
- What is the role of...?
- Which structure...?
- What causes...?
- What happens when...?

Avoid unnecessarily sophisticated or artificial wording.

BAD:
"What specific anatomical landmark marks the termination of the human
digestive tract's posterior opening?"

BETTER:
"Which structure marks the posterior end of the human alimentary canal?"

BAD:
"What specific structural role does water fulfill regarding internal
physiological homeostasis?"

BETTER:
"What role does water play in the human body?"

Do not make a question harder merely by using complicated vocabulary.

Avoid meta phrases such as:

- "according to the provided text"
- "according to the passage"
- "mentioned in the text"
- "based on the information given"
- "as stated above"

The question should stand naturally on its own.


==================================================
6. QUESTION-ANSWER ALIGNMENT
==================================================

The question must ask exactly what the correct answer provides.

The grammatical and conceptual relationship between the question and
answer must be correct.

Examples:

"Why does X happen?"
→ The answer must provide a reason or cause.

"What is X?"
→ The answer must provide a definition or identity.

"What is the function of X?"
→ The answer must provide a function.

"Which feature characterizes X?"
→ The answer must provide a characteristic.

"Where is X located?"
→ The answer must provide a location.

"How does X occur?"
→ The answer must describe a mechanism or process.

"Which structure is X?"
→ The answer must identify the structure.

Do NOT ask a "why" question when the answer only gives a definition,
characteristic, location, or example.

Do NOT ask a "how" question when the answer only identifies something.

Do NOT ask for a function when the answer describes a structure.

The question and correct answer must be logically aligned.


==================================================
7. OPTIONS MUST HAVE THE SAME CATEGORY
==================================================

Before writing the options, identify the CATEGORY of the correct answer.

All four options must belong to the same conceptual category.

Examples:

If the question asks:

"Why do herbivores have longer alimentary canals?"

Then all four options must be possible reasons for longer alimentary
canals.

If the question asks:

"What is the function of the pancreas?"

Then all four options must be functions of the pancreas.

If the question asks:

"Which structure is part of the alimentary canal?"

Then all four options must be anatomical structures that could reasonably
be considered parts of the alimentary canal.

If the question asks:

"What causes disease X?"

Then all four options must be possible causes.

Do NOT mix categories.

BAD:

Q: Why do herbivores have longer alimentary canals?

A. To provide a larger absorptive surface area
B. Humans are mostly omnivores
C. Carbohydrates provide energy
D. Water is reabsorbed in the large intestine

B, C and D belong to unrelated categories.

BETTER:

Q: Why do herbivores have longer alimentary canals?

A. To provide a larger absorptive surface for processing plant material
B. To provide additional space for processing plant material
C. To support the digestion of difficult-to-digest plant material
D. To allow more time for nutrients from plant food to be absorbed

All four options belong to the same conceptual category.


==================================================
8. DISTRACTOR REQUIREMENTS
==================================================

Each question must have exactly three incorrect options.

Every wrong option should satisfy these requirements:

1. SAME CATEGORY
   It belongs to the same conceptual category as the correct answer.

2. PLAUSIBLE
   It should be believable to a student who understands the topic.

3. SOURCE-GROUNDED
   Prefer facts or concepts supported by the provided source.

4. RELEVANT
   It should have a meaningful conceptual relationship to the question.

5. NOT OBVIOUSLY WRONG
   A student should not be able to eliminate it simply because it is
   absurd, unrelated, or common-sense nonsense.

6. NOT DUPLICATE
   It must not simply restate another option.

Whenever possible, wrong options should be TRUE-BUT-WRONG:
they should represent information that is factually supported by the
source but does not correctly answer the particular question.

If three strong source-supported distractors are not available, use
closely related concepts explicitly discussed in the source.

NEVER invent scientific facts merely to create distractors.

If three genuinely plausible distractors cannot be created without
inventing information, SKIP that question.


==================================================
9. FORBIDDEN DISTRACTOR PATTERNS
==================================================

Do NOT create:

- Absurd options
- Obviously false options
- Unrelated scientific concepts
- Random organs or diseases
- Made-up scientific facts
- Options from different conceptual categories
- Options that can be eliminated through common sense alone
- Options that are much longer or shorter in a way that reveals the answer
- Options that repeat the wording of the correct answer
- Multiple options that could reasonably be considered correct


==================================================
10. AVOID ANSWER AMBIGUITY
==================================================

There must be exactly ONE defensible correct answer.

Before finalizing each question, check:

- Could another option also be correct?
- Does the source clearly support only one answer?
- Are any two options partially correct?
- Does the question wording make the intended answer unambiguous?

If more than one option could reasonably be correct, rewrite the
question or options.

If the ambiguity cannot be resolved from the source, skip the question.


==================================================
11. PARAPHRASING
==================================================

Test understanding rather than memorization of exact phrases.

Paraphrase the question and answer options naturally.

Do NOT unnecessarily copy sentences word-for-word from the source.

However, scientific terminology, anatomical names, species names,
disease names, and other necessary technical terms may remain unchanged.

Do not change technical terminology merely for the sake of paraphrasing.


==================================================
12. DIFFICULTY
==================================================

Assign each question one of:

- EASY
- MEDIUM
- HARD

Use these definitions:

EASY:
A directly stated fact, definition, identification, or simple recall.

MEDIUM:
Requires distinguishing between related concepts, connecting two
facts, understanding a described process, or selecting between plausible
alternatives.

HARD:
Requires applying or synthesizing multiple facts from the source or
reasoning through a biological relationship.

IMPORTANT:

Do NOT label a question HARD merely because its wording is complicated.

Do NOT label a question MEDIUM merely because it contains scientific
terminology.

Do NOT artificially increase difficulty.

Difficulty should reflect the cognitive demand of the question.


==================================================
14. FINAL QUALITY CHECK
==================================================

Before returning the JSON, internally verify EVERY question.

For each question check:

1. SOURCE SUPPORT
   Is the question answerable using only the provided source?

2. QUESTION-ANSWER ALIGNMENT
   Does the question ask exactly what the correct answer provides?

3. SINGLE CORRECT ANSWER
   Is exactly one option clearly correct?

4. SAME CATEGORY
   Do all four options belong to the same conceptual category?

5. DISTRACTOR QUALITY
   Are the three wrong options plausible and relevant?

6. DISTRACTOR SOURCE SUPPORT
   Are the wrong options grounded in the source whenever possible?

7. NO COMMON-SENSE ELIMINATION
   Could a student eliminate a distractor without knowing the topic?

8. NO INVENTED INFORMATION
   Did you introduce anything that is not supported by the source?

9. NATURAL LANGUAGE
   Is the question clear and natural rather than unnecessarily complex?

10. DIFFICULTY
    Does the difficulty reflect the actual cognitive demand?

11. REMARK
    Does every correct option have a NON-EMPTY 2-3 sentence remark?
    Does the remark contain source-supported facts AND reasoning?
    Does it explain the concept instead of merely repeating the answer?
    Are all incorrect options given an empty remarks field?

12. COVERAGE
    Does the question test a distinct concept rather than unnecessarily
    repeating another question?

If any check fails, rewrite the question/options.

If it still cannot satisfy all requirements, remove that question.

Never sacrifice question quality merely to increase the number of MCQs.

==================================================
FINAL RULE — WHAT GOES IN "remarks"
==================================================

The "remarks" field is REQUIRED for the correct option only.
It is an EMPTY STRING for wrong options.

"remarks" must contain TWO parts, in order:

  PART 1 — FACT
    One sentence stating a factual detail from the source about
    the concept being tested.

  PART 2 — REASONING
    One or two sentences explaining why that fact makes this option
    correct, using relationships or mechanisms from the source.

Total: 2-3 sentences. Minimum 25 words.

If your remark is fewer than 25 words, it is TOO SHORT.
If your remark is a label like "Correct definition", it is WRONG.
If your remark just restates the answer, it is WRONG.

EXAMPLES:

 WRONG (label):
  "Correct physiological role."

 WRONG (restates answer):
  "Carbohydrates serve as the chief energy providers in human food."

 CORRECT (fact + reasoning):
  "Carbohydrates act as the body's primary immediate fuel source.
   They are broken down into glucose, which cells use for energy.
   Unlike proteins which build tissue or fats which store energy,
   carbohydrates are the primary energy source for cellular work."

 WRONG (label):
  "Correct mechanism."

 WRONG (restates answer):
  "Marasmus results from protein-calorie undernutrition."

 CORRECT (fact + reasoning):
  "Marasmus is associated with severe protein-calorie malnutrition
   and energy deficiency. Inadequate intake of protein and calories
   deprives the body of the nutrients and energy needed to maintain
   normal tissues, leading to severe wasting."

REMINDER: 
- 2-3 sentences.
- Minimum 25 words.
- Fact first, then reasoning.
- Only the correct option has a remark.

==================================================
15. IMPORTANT PRINCIPLE
==================================================

The goal is NOT to generate as many questions as possible.

The goal is to generate the maximum number of HIGH-QUALITY,
SOURCE-GROUNDED MCQs that can be reliably supported by the provided
Zoology content.

A smaller number of strong questions is better than a larger number
of weak, repetitive, ambiguous, or invented questions.

==================================================
17. FORBIDDEN QUESTION TYPES
==================================================

Do NOT write questions about:

- Titles, designations, or nicknames
  ✗ "What title is given to X?"
  ✗ "What is X also called?"
  ✗ "What name is used for X?"

- Metaphors or figures of speech
  ✗ "What does the text call X?"

- Textbook structure or headings
  ✗ "Which section covers X?"

These test memory of wording, not understanding.

Instead, ask about:
✓ Functions: "What is the role of X?"
✓ Causes: "Why does X happen?"
✓ Mechanisms: "How does X work?"
✓ Comparisons: "How does X differ from Y?"

Remove using as mentioned in the text, from the given text
"""
class Option(BaseModel):
    option: str
    remarks: str 
    is_correct: bool
    
class MCQ(BaseModel):
    question: str
    difficulty: Literal["EASY", "MEDIUM", "HARD"]
    status: Literal["PENDING"] = "PENDING"
    options: list[Option] = Field(min_length=4, max_length=4)

class MCQList(BaseModel):
    mcqs: list[MCQ] = Field(
        description="MCQs generated. May be empty if text has no testable content.",
        max_length=10,
    )
    
def generate_mcqs(chunk_text):
    """Send chunk to Gemini, return list of MCQs."""
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=f"Generate UP TO 10 MCQs from this text. Base the count on content quality.\n\n{chunk_text}",
        config=types.GenerateContentConfig(
            system_instruction=SYSTEM_PROMPT,
            response_mime_type= 'application/json',
            response_schema=MCQList,
        )
                              
    )
    return json.loads(response.text)["mcqs"]


with open("chapter1_chunks.json", "r", encoding="utf-8") as f:
    chunks = json.load(f)

print(f"Total chunks: {len(chunks)}")


candidates= []
for i, chunk in enumerate(chunks):
    text = chunk["text"]
    print(f"[{i+1}/{len(chunks)}] {chunk['id']} ({len(text)} chars)")
    try:
        mcqs = generate_mcqs(text)
        for q in mcqs:
            candidates.append({
                "chunk_id": chunk["id"],
                **q,
            })

        if mcqs:
            print(f"   {len(mcqs)} candidates")
        else:
            print(f"  0 candidates (no testable content)")

    except Exception as e:
        print(f"  ✗ Error: {e}")
    with open("mcqs_candidates.json", "w", encoding="utf-8") as f:
        json.dump(candidates, f, ensure_ascii=False, indent=2)

    time.sleep(10)   # rate limit

print(f"\n Total candidates: {len(candidates)}")
print(f"  Saved to mcqs_candidates.json")
```

```python
import json
from collections import Counter
from pydantic import BaseModel

with open("mcqs_candidates.json", encoding="utf-8") as f:
    candidates = json.load(f)

questions = []
for entry in candidates:
    questions.extend(entry.get("mcqs", [entry]))

pool = {"EASY": [], "MEDIUM": [], "HARD": []}
for q in questions:
    pool[q["difficulty"]].append(q)

print(f"EASY: {len(pool['EASY'])}, MEDIUM: {len(pool['MEDIUM'])}, HARD: {len(pool['HARD'])}")

targets = {"EASY": 25, "MEDIUM": 20, "HARD": 5}

class Pick(BaseModel):
    selected_indices: list[int]

# Concept context — all question stems so Gemini knows what repeats
all_stems = "\n".join(f"- {q['question']}" for q in questions)

selected = []
shortfall = 0

for level in ["EASY", "MEDIUM", "HARD"]:
    tier = pool[level]
    target = targets[level]

    if len(tier) == 0:
        shortfall += target
        continue

    if len(tier) <= target:
        selected.extend(tier)
        if len(tier) < target:
            shortfall += target - len(tier)
        continue

    # Only send THIS tier's questions
    tier_text = ""
    for i, q in enumerate(tier):
        opts = "\n".join(
            f"    {'✓' if o['is_correct'] else ' '} {o['option']}"
            for o in q["options"]
        )
        tier_text += f"\n[{i}] Q: {q['question']}\n{opts}\n"

    prompt = f"""Select the BEST {target} {level} questions from the {len(tier)} below.

IMPORTANCE — a question is more important if its concept ALSO appears in
other parts of the chapter. The list of ALL question stems from the pool
is given below. Prefer questions whose concepts repeat there.

QUALITY — same-category options, plausible distractors, one clear answer,
short options (under 10 words). No titles or metaphor questions.

Return JSON: {{"selected_indices": [0, 3, 7, ...]}}
Select EXACTLY {target} questions.

All concepts in the chapter (use for repetition judgement):
{all_stems}

═══════════════════════════════════════════
{level} QUESTIONS TO PICK FROM:
═══════════════════════════════════════════
{tier_text}
"""

    for model in ["gemini-3.1-flash-lite", "gemini-3.5-flash-lite"]:
        try:
            r = client.models.generate_content(
                model=model,
                contents=prompt,
                config=types.GenerateContentConfig(
                    response_mime_type="application/json",
                    response_schema=Pick,
                ),
            )
            idx = json.loads(r.text)["selected_indices"]
            picked = [tier[i] for i in idx if i < len(tier)]
            selected.extend(picked)
            print(f"  {level}: picked {len(picked)} with {model}")
            if len(picked) < target:
                shortfall += target - len(picked)
            break
        except Exception as e:
            if "503" in str(e) or "UNAVAILABLE" in str(e) or "429" in str(e):
                continue
            raise

# Fill shortfall from MEDIUM
if shortfall > 0 and len(selected) < 50:
    used = set(id(q) for q in selected)
    remaining = [q for q in pool["MEDIUM"] if id(q) not in used]
    selected.extend(remaining[:shortfall])

selected = selected[:50]
clean = []
for q in selected:
    clean.append({
        "question": q["question"],
        "difficulty": q["difficulty"],
        "status": q.get("status", "PENDING"),
        "options": [
            {
                "option": o["option"],
                "remarks": o.get("remarks", ""),
                "is_correct": o["is_correct"],
            }
            for o in q["options"]
        ],
    })

with open("mcqs_50.json", "w", encoding="utf-8") as f:
    json.dump(clean, f, ensure_ascii=False, indent=2)

diff = Counter(q["difficulty"] for q in clean)
print(f"\nFinal: {len(clean)}")
for d in ["EASY", "MEDIUM", "HARD"]:
    print(f"  {d}: {diff.get(d, 0)}")
print("\n Saved mcqs_50.json")
```

```python
import os, json, time
from dotenv import load_dotenv
from google import genai
from google.genai import types
from PIL import Image, ImageDraw, ImageFont
from textwrap import wrap

load_dotenv()
api_key = os.getenv("GEMINI_API_KEY") or os.getenv("GOOGLE_API_KEY")
client = genai.Client(api_key=api_key)

IMAGE_FOLDER = "images"
OUTPUT_FOLDER = "images_with_captions"
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

PROMPT = """Describe this image in detail in 4-6 sentences.
Include: type of image, main subject, all visible labels/text,
key visual elements, and relationships shown.
Be factual. Do not invent anything not visible."""

files = sorted(f for f in os.listdir(IMAGE_FOLDER)
               if f.lower().endswith((".png", ".jpg", ".jpeg", ".webp")))

for fname in files:
    path = os.path.join(IMAGE_FOLDER, fname)
    print(f"Captioning {fname} ...")

    with open(path, "rb") as f:
        data = f.read()

    ext = fname.lower().rsplit(".", 1)[-1]
    mime = {"png": "image/png"}[ext]

    r = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=[PROMPT, types.Part.from_bytes(data=data, mime_type=mime)]
    )
    caption = r.text.strip()


    img = Image.open(path).convert("RGB")
    W, H = img.size
    font_size = max(16, W // 60)
    try:
        font = ImageFont.truetype("arial.ttf", font_size)
    except:
        font = ImageFont.load_default()


    wrapped = []
    for line in caption.split("\n"):
        wrapped += wrap(line, width=max(40, W // (font_size // 2)))

    line_h = font_size + 6
    pad = 20
    cap_h = len(wrapped) * line_h + pad * 2


    new_img = Image.new("RGB", (W, H + cap_h), "white")
    new_img.paste(img, (0, 0))

    draw = ImageDraw.Draw(new_img)
    y = H + pad
    for line in wrapped:
        draw.text((pad, y), line, fill="black", font=font)
        y += line_h

    out_path = os.path.join(OUTPUT_FOLDER, fname)
    new_img.save(out_path)
    print(f"  ✓ saved {out_path}")

    time.sleep(4)

print("Done.")
```

```python

import fitz
import os
import json
import time
import shutil
from dotenv import load_dotenv
from google import genai
from google.genai import types

# --- 1. Setup ---
load_dotenv()
api_key = os.getenv("GEMINI_API_KEY") or os.getenv("GOOGLE_API_KEY")
client = genai.Client(api_key=api_key)

PDF_FILE = "Zoology.pdf"
IMAGE_FOLDER = "extracted_diagrams"

# --- 2. Extract Diagram Crops (Smart Filtering) ---
print("Step 1: Finding and extracting diagrams (Pages 20-57)...")
if os.path.exists(IMAGE_FOLDER):
    shutil.rmtree(IMAGE_FOLDER)
os.makedirs(IMAGE_FOLDER, exist_ok=True)

doc = fitz.open(PDF_FILE)
extracted_images = []
start_page = 20
end_page = 56 

for page_num in range(start_page, end_page + 1):
    if page_num >= len(doc):
        break
        
    page = doc[page_num]
    drawings = page.get_drawings()
    images = page.get_images(full=True)
    
    rects = []
    
    # Get image bounds
    for img in images:
        bbox = page.get_image_bbox(img)
        if bbox.width < 500 and bbox.height < 500:
            rects.append(bbox)
        
    # Get drawing bounds (STRICTER FILTER)
    for d in drawings:
        rect = d["rect"]
        # ONLY keep shapes between 50x50 and 350x350
        # This ignores tiny dots AND massive full-page borders
        if 50 < rect.width < 350 and 50 < rect.height < 350:
            rects.append(rect)
            
    if not rects:
        print(f"  Page {page_num + 1}: No diagrams found (Skipped)")
        continue 
        
    # Merge all rects into one big bounding box
    merged_rect = rects[0]
    for r in rects[1:]:
        merged_rect |= r  
        
    # Safety check: If the box is still almost the whole page (A4 is 595x842), skip it
    if merged_rect.width > 500 or merged_rect.height > 700:
        print(f"  Page {page_num + 1}: Box too large, likely a border (Skipped)")
        continue
        
    # Add a tiny margin
    merged_rect = merged_rect + (-10, -10, 10, 10)
    
    # Crop and save
    pix = page.get_pixmap(clip=merged_rect, dpi=200)
    img_filename = f"{IMAGE_FOLDER}/diagram_p{page_num+1:03d}.png"
    pix.save(img_filename)
    
    extracted_images.append({
        "page": page_num + 1,
        "file": img_filename
    })
    print(f"  Page {page_num + 1}: Extracted (Box: {int(merged_rect.width)}x{int(merged_rect.height)})")

print(f"\n   Found {len(extracted_images)} diagrams total.")

# --- 3. Caption the Cropped Diagrams ---
print("\nStep 2: Captioning cropped diagrams...")
PROMPT = """Describe this Zoology diagram in 4-6 sentences.
Include: what the diagram is showing, all visible labels, and the relationships between parts.
Be factual. Do not invent anything not visible."""

dataset = []

for item in extracted_images:
    print(f"Captioning page {item['page']}...", end=" ")
    
    with open(item["file"], "rb") as f:
        img_bytes = f.read()
        
    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash-lite",
            contents=[PROMPT, types.Part.from_bytes(data=img_bytes, mime_type="image/png")]
        )
        
        caption = response.text.strip()
        print("✓")
        
        dataset.append({
            "page": item["page"],
            "image_file": item["file"],
            "caption": caption
        })
        
    except Exception as e:
        print(f" Error: {e}")
        
    time.sleep(4)

# --- 4. Save ---
output_file = "zoology_diagrams.json"
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(dataset, f, indent=2, ensure_ascii=False)

print(f"\n Saved {len(dataset)} cropped diagrams + captions to {output_file}")
```
