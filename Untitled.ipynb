```python
import pymupdf

doc = pymupdf.open("Zoology.pdf")
print("Total pages:", len(doc))

text = ""

for page_number in range(20, 57):
    page = doc[page_number]
    for block in page.get_text("dict")["blocks"]:
        if block["type"] != 0:
            continue
        for line in block["lines"]:
            for span in line["spans"]:
                size = span["size"]
                font = span["font"]

                if "CenturyGothic" in font:
                    continue
                if size <= 10.0:
                    continue
                if size >= 20:
                    continue
                text += span["text"]
            text += "\n"

doc.close()

print("Characters extracted:", len(text))
print(text[:1000])
```

```python
import pymupdf

doc = pymupdf.open("Zoology.pdf")
print("Total pages:", len(doc))

text = ""

for pno in range(20, 57):
    page = doc[pno]

    page_h = page.rect.height

    for block in page.get_text("dict")["blocks"]:
        if block["type"] != 0:
            continue
        for line in block["lines"]:
            line_text = ""
            for span in line["spans"]:
                line_text += span["text"]
            line_text = " ".join(line_text.split())

            if not line_text:
                continue

            x0, y0, x1, y1 = line["bbox"]

            # Position: header band
            if y0 < 0.08 * page_h:
                continue
            # Position: footer band
            if y1 > 0.92 * page_h:
                continue

            # Figure caption
            if line_text.startswith("Figure"):
                continue


            text += line_text + "\n"

doc.close()

# Whitespace cleanup
lines = [" ".join(l.split()) for l in text.splitlines() if l.strip()]
cleaned = "\n".join(lines)
print(cleaned[:4000])
```

```python
import pymupdf4llm
md = pymupdf4llm.to_markdown("Zoology.pdf", pages=list(range(20, 57)))
# lines = md.splitlines()
print(md[:5000])
```

```python
cleaned = []
lines = md.splitlines()
in_picture = False

for line in lines:
    s = line.strip()
    if "<!-- Start of picture text -->" in s:
        in_picture = True
        continue
    if "<!-- End of picture text -->" in s:
        in_picture = False
        continue
    if in_picture:
        continue
    cleaned.append(line)

lines = cleaned
print("After step 1:", len(lines), "lines")
print("\n".join(lines[:30]))
```

```python
cleaned = []
blank_count = 0

for line in lines:
    s = line.strip()
    content = s.lstrip("#").strip()

    # Step 2 — drop page numbers
    if content.isdigit() and len(content) <= 3:
        continue

    # Step 3a — drop short non-heading lines
    if len(content) < 15 and not s.startswith("#"):
        continue

    # Step 3b — collapse blank lines
    if not s:
        blank_count += 1
        if blank_count <= 1:
            cleaned.append("")
        continue

    blank_count = 0
    cleaned.append(line)

lines = cleaned

print("\n".join(lines[:40]))
```

```python
cleaned = []
for line in lines:
    if "~~" in line:
        continue
    cleaned.append(line)

lines = cleaned
print("After step 4:", len(lines), "lines")
print("\n".join(lines[:40]))
```

```python
cleaned = []
for line in lines:
    s = line.strip()

    # Drop bold-only running header
    if (s.startswith("**") and s.endswith("**")
        and s.count("**") == 2
        and not s.startswith("#")
        and not any(c.isdigit() for c in s)):
        continue

    if s.startswith("Figure ") and len(s) < 100:
        continue

    if line.strip() == "Human Anatomy and Physiology - I" :
        continue
         
    if line.strip() == "Zoology":
        continue

    cleaned.append(line)

lines = cleaned
print("After bold-header step:", len(lines), "lines")
print("\n".join(lines[:60]))
```

```python
cleaned = []
for line in lines:
    s = line

    # Remove heading markers (### at start)
    while s.startswith("#"):
        s = s[1:]
    s = s.lstrip()   # remove leading space

    # Remove bold markers (**...**)
    s = s.replace("**", "")

    # Remove italic markers (_..._)
    s = s.replace("_", "")

    # Remove stray asterisks (*)
    s = s.replace("*", "")

    cleaned.append(s)

lines = cleaned
print("After removing markers:", len(lines), "lines")
print("\n".join(lines[:40]))

with open("chapter1_clean.md", "w", encoding="utf-8") as f:
    f.write("\n".join(lines))

print("Saved chapter1_clean.md")
print("Lines:", len(lines))
print("Chars:", len("\n".join(lines)))
```
