```python
import pymupdf

# Open PDF
doc = pymupdf.open("Zoology.pdf")

print("Total pages:", len(doc))

text = ""

for page_number in range(21,57):
    text += doc[page_number].get_text()+"\n"

doc.close()

print("Characters extracted:", len(text))
print(text[:3000])
```
